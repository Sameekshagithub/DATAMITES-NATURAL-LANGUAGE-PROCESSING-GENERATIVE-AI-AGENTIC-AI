# HuggingFace — Encoder Models (BERT)

**Encoder models READ and UNDERSTAND text.**
They do NOT generate new text — they are great at:
- Sentiment Analysis (positive / negative)
- Named Entity Recognition (find names, places, dates)
- Text Classification (spam or not spam)
- Sentence similarity / search

Famous encoder models: **BERT**, **RoBERTa**, **DistilBERT**

---
We will use **HuggingFace `pipeline`** — the easiest way to use these models.
One line loads a model. One line runs it.

```python
from transformers import pipeline
```

---
## Task 1 — Sentiment Analysis

**What it does:** Reads a sentence and tells you if it is POSITIVE or NEGATIVE.

**Model used:** `distilbert-base-uncased-finetuned-sst-2-english`
- DistilBERT = smaller, faster version of BERT
- Fine-tuned on SST-2 = a dataset of movie reviews labelled positive/negative

```python
sentiment = pipeline("sentiment-analysis")

sentences = [
    "I absolutely loved this movie!",
    "The food was cold and the service was terrible.",
    "It was neither good nor bad.",
]

results = sentiment(sentences)

for sentence, result in zip(sentences, results):
    print(f"Input   : {sentence}")
    print(f"Result  : {result['label']}  (confidence: {result['score']:.2%})")
    print()
```

---
## Task 2 — Named Entity Recognition (NER)

**What it does:** Reads a sentence and labels each word:
- `PER` = person name
- `ORG` = organisation
- `LOC` = location / place
- `MISC` = other named entity

**Real-world use:** Medical records, legal documents, news articles

```python
ner = pipeline("ner", aggregation_strategy="simple")

text = "Elon Musk founded SpaceX in Hawthorne, California in 2002."

entities = ner(text)

print(f"Input: {text}")
print()
print("Entities found:")
for e in entities:
    print(f"  [{e['entity_group']:4s}]  '{e['word']}'   (confidence: {e['score']:.2%})")
```

---
## Task 3 — Question Answering

**What it does:** Given a paragraph (context) and a question, the encoder finds the answer **inside the paragraph**.

It does NOT make up an answer — it highlights the exact span of text.

**Real-world use:** Customer support bots, document search, FAQs

```python
qa = pipeline("question-answering")

context = """
The Eiffel Tower is a wrought-iron lattice tower on the Champ de Mars in Paris, France.
It was constructed from 1887 to 1889 as the centerpiece of the 1889 World's Fair.
It is named after the engineer Gustave Eiffel, whose company designed and built the tower.
"""

questions = [
    "Where is the Eiffel Tower located?",
    "When was the Eiffel Tower built?",
    "Who designed the Eiffel Tower?",
]

for question in questions:
    result = qa(question=question, context=context)
    print(f"Q: {question}")
    print(f"A: {result['answer']}   (confidence: {result['score']:.2%})")
    print()
```

---
## Task 4 — Zero-Shot Classification

**What it does:** Classifies text into categories **you define yourself** — no training needed!

The encoder understands meaning deeply enough to match text to any label you give it.

**Real-world use:** Routing customer emails, tagging news articles, content moderation

```python
classifier = pipeline("zero-shot-classification")

text   = "The new iPhone 16 has an impressive camera and long battery life."
labels = ["technology", "sports", "politics", "food", "health"]

result = classifier(text, candidate_labels=labels)

print(f"Input : {text}")
print()
print("Category scores:")
for label, score in zip(result["labels"], result["scores"]):
    bar = "█" * int(score * 40)
    print(f"  {label:12s} {score:.2%}  {bar}")
```

---
## Task 5 — Feature Extraction (Get Word Vectors)

**What it does:** Returns the raw number vectors (embeddings) that BERT produces for each word.
These vectors capture meaning — similar words will have similar vectors.

**Real-world use:** Semantic search engines, recommendation systems, clustering

```python
import torch

features = pipeline("feature-extraction", return_tensors=True)

sentences = ["Paris is beautiful.", "France is lovely."]

vecs = [features(s)[0].mean(dim=0) for s in sentences]  # average over all tokens

# Cosine similarity — 1.0 = identical meaning, 0.0 = unrelated
similarity = torch.nn.functional.cosine_similarity(vecs[0], vecs[1], dim=0)

print(f"Sentence 1: '{sentences[0]}'")
print(f"Sentence 2: '{sentences[1]}'")
print(f"\nEach sentence → vector of {vecs[0].shape[0]} numbers (BERT hidden size)")
print(f"\nSemantic similarity: {similarity.item():.4f}")
print("(1.0 = identical meaning,  0.0 = no relation)")
```

## Summary

| Task | Pipeline name | What it returns |
|---|---|---|
| Sentiment Analysis | `sentiment-analysis` | POSITIVE / NEGATIVE + score |
| Named Entity Recognition | `ner` | Words labelled as PER, ORG, LOC |
| Question Answering | `question-answering` | Exact answer span from context |
| Zero-Shot Classification | `zero-shot-classification` | Score for each label you define |

All of these use **Encoder models** — they READ and UNDERSTAND, never generate.
