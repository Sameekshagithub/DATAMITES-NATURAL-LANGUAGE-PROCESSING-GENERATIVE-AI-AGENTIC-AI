## Step 2: Import Libraries

```python
import os
import re
import string
import emoji
import contractions
import nltk

from bs4 import BeautifulSoup
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
```

## Step 3: Download NLTK Data (one-time)

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

## Step 4: File Handling — Create a Sample Text File

Write some messy raw text (with HTML tags, a URL, emojis, punctuation, numbers) into a `.txt` file.


```python
raw_text = """<p>Hello World!! 😊 Visit https://www.example.com today!!! 
I have 2 cats & 3 dogs. #NLP @python Machine Learning is amazing... 🚀🔥
Can't wait to learn Natural Language Processing!!!</p>"""

with open("raw_text.txt", "w") as file:
    file.write(raw_text)

print("File written successfully!")
print(os.path.exists("raw_text.txt"))
```

## Step 5: Read the File Back

```python
with open("raw_text.txt", "r") as file:
    text = file.read()

print("="*60)
print("Original Text:\n")
print(text)
```

## Step 6: Text Cleaning

Removing HTML tags, URLs, emojis, punctuation, numbers, and extra whitespace.


```python
# STEP A: Remove HTML tags
text = BeautifulSoup(text, "html.parser").get_text()
print("After removing HTML tags:\n")
print(text)
```

```python
# STEP B: Lowercase
text = text.lower()
print("After lowercase:\n")
print(text)
```

```python
# STEP C: Expand contractions (can't -> cannot)
text = contractions.fix(text)
print("After expanding contractions:\n")
print(text)
```

```python
# STEP D: Remove URLs
text = re.sub(r"http\S+|www\S+", "", text)
print("After removing URLs:\n")
print(text)
```

```python
# STEP E: Remove emojis
text = emoji.replace_emoji(text, replace="")
print("After removing emojis:\n")
print(text)
```

```python
# STEP F: Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))
print("After removing punctuation:\n")
print(text)
```

```python
# STEP G: Remove numbers
text = re.sub(r"\d+", "", text)
print("After removing numbers:\n")
print(text)
```

```python
# STEP H: Remove extra whitespace
text = re.sub(r"\s+", " ", text).strip()
print("Final Cleaned Text:\n")
print(text)
```

## Step 7: Tokenization

```python
word_tokens = word_tokenize(text)
print("Word Tokens:\n")
print(word_tokens)
```

```python
sent_tokens = sent_tokenize(text)
print("Sentence Tokens:\n")
print(sent_tokens)
```

## Step 8: Stopword Removal

```python
stop_words = set(stopwords.words("english"))
filtered_tokens = [word for word in word_tokens if word not in stop_words]

print("Tokens after removing stopwords:\n")
print(filtered_tokens)
```

## Step 9: Lemmatization

Lemmatization is done with POS tags so words are reduced to their correct dictionary form
(e.g. "running" -> "run", "better" -> "good").


```python
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return "a"   # adjective
    elif tag.startswith("V"):
        return "v"   # verb
    elif tag.startswith("N"):
        return "n"   # noun
    elif tag.startswith("R"):
        return "r"   # adverb
    else:
        return "n"   # default to noun
```

```python
pos_tags = pos_tag(filtered_tokens)

lemmatized_tokens = [
    lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    for word, tag in pos_tags
]

print("Lemmatized Tokens:\n")
print(lemmatized_tokens)
```

## Step 10: Save the Cleaned Text to a New File

```python
cleaned_text = " ".join(lemmatized_tokens)

with open("cleaned_text.txt", "w") as file:
    file.write(cleaned_text)

print("Cleaned text saved to cleaned_text.txt")
print(os.path.exists("cleaned_text.txt"))

with open("cleaned_text.txt", "r") as file:
    print("\nFinal Output:\n")
    print(file.read())
```
