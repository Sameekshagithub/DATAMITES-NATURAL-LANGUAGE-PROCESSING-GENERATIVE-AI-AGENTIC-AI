```python
def num_to_word(sen):
    sentence = re.sub("\d",q.number_to_words(sen),sen)
    return sentence
```

```python
num_to_word('i have 32 chocolates')
```

```python
q.number_to_words('i have 32 chocolates')
```

```python
def convert_to_num(text):
    new_list = []
    mylist = text.split()
    for i in mylist:
        if i.isdigit():
            new_list.append(q.number_to_words(i))
        else:
            new_list.append(i)
    return " ".join(new_list)
```

```python
#importing libraries
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# here we define chunking function with text and regular
# expressions representing grammar as parameter
def chunking(text, grammar):
    word_tokens = word_tokenize(text)

    # label words with pos
    word_pos = pos_tag(word_tokens)

    # create chunk parser using grammar
    chunkParser = nltk.RegexpParser(grammar)

    # test it on the list of word tokens with tagged pos
    tree = chunkParser.parse(word_pos)

    for subtree in tree.subtrees():
        print(subtree)
    #tree.draw()

sentence = 'the little red parrot is flying in the sky'
"""NP: This indicates that we are defining a noun phrase (NP).
{ ... }: This curly brace notation is used to enclose the pattern definition.
<DT>?: This part specifies that an optional determiner (DT) may appear at the beginning of the noun phrase.
<JJ>*: This indicates that zero or more adjectives (JJ) may occur after the determiner (if present) and before the noun.
<NN>: This specifies that the noun (NN) must be present as the final element of the noun phrase."""

grammar = "NP: {<DT>?<JJ>*<NN>}"
chunking(sentence, grammar)
```
