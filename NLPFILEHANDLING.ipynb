```python
from PyPDF2 import PdfReader

reader=PdfReader("sample.pdf")

print(reader)
```

```python
reader=PdfReader("sample.pdf")
total_pages=len(reader.pages)
print(total_pages)
```

```python
from PyPDF2 import PdfReader

reader=PdfReader("sample.pdf")

page=reader.pages[0]

text=page.extract_text()

print(text)
```

```python
from PyPDF2 import PdfReader

reader=PdfReader("sample.pdf")

for pages in reader.pages:
  text=page.extract_text()
  print(text)
```

```python
reader=PdfReader("sample.pdf")
writer=PdfWriter()

for page in reader.pages:
  writer.add_page(page)

with open("copy.pdf","wb") as file:
  writer.write(file)
```

```python
reader=PdfReader("sample.pdf")
writer=PdfWriter()
writer.add_page(reader.pages[0])
with open("Page1.pdf","wb") as file:
  writer.write(file)
```

```python
reader=PdfReader("sample.pdf")
writer=PdfWriter()
writer.add_page(reader.pages[1])
with open("Page2.pdf","wb") as file:
  writer.write(file)
```

```python
#merge
from PyPDF2 import PdfMerger

merger = PdfMerger()

merger.append("Page1.pdf")
merger.append("Page2.pdf")

merger.write("Merged.pdf")
merger.close()
```
