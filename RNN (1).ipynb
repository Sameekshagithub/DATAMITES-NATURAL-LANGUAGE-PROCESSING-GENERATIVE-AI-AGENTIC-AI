# RNN --- RECURRENT NEURAL NETWORK

```python
# 1️⃣ Imports
from tensorflow.keras.models import Sequential # BUILD THE MODEL LAYER BY LAYER
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense #Embedding - convertig words into numbers
# Simple RNN - RNN layer, Dense - output layer
from tensorflow.keras.preprocessing.text import Tokenizer # converting sentence into token
from tensorflow.keras.preprocessing.sequence import pad_sequences # makes all input seq same length
import numpy as np # used to work on arrays
```

```python
# 2️⃣ Sample sentences
texts = ["I love NLP", "Deep learning is fun", "ANN is not good for sequences"]
labels = [1, 1, 0]  # Example: 1 = Positive, 0 = Negative
```

```python
# 3️⃣ Tokenize text
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
padded_seq = pad_sequences(sequences, padding='post')
```

```python
sequences
```

```python
padded_seq
```

```python
# 4️⃣ Build RNN Model
model = Sequential()
model.add(Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=8))
model.add(SimpleRNN(16))  # RNN layer
model.add(Dense(1, activation='sigmoid'))  # Binary classification
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# 5️⃣ Train the model
model.fit(padded_seq, np.array(labels), epochs=20, verbose=1)
```

```python
# 6️⃣ Test with new sentence
test_text = ["I don't like deep learning"]
test_seq = tokenizer.texts_to_sequences(test_text)
test_pad = pad_sequences(test_seq, maxlen=padded_seq.shape[1], padding='post')
prediction = model.predict(test_pad)
```

```python
print("Prediction:", "Positive" if prediction[0][0] > 0.5 else "Negative")
```
