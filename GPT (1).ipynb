# HuggingFace — Decoder Models (GPT-2)

**Decoder models WRITE and GENERATE text.**
They predict the next word, one at a time, until done.

They are great at:
- Text generation / completion
- Chatbots
- Code generation
- Story writing
- Autocomplete

Famous decoder models: **GPT-2**, **GPT-3**, **GPT-4**, **LLaMA**, **Mistral**

---
We will use **GPT-2** (free and open) via HuggingFace `pipeline`.

```python
from transformers import pipeline
```

---
## Task 1 — Text Completion

**What it does:** You give the model the start of a sentence. It continues writing.

This is exactly how ChatGPT, Copilot, and phone autocomplete work — just with much bigger models.

`max_new_tokens` = how many new words to generate  
`num_return_sequences` = how many different completions to show

```python
generator = pipeline("text-generation", model="gpt2")

prompt = "Artificial intelligence will change the world by"

results = generator(prompt, max_new_tokens=40, num_return_sequences=3, truncation=True)

print(f"Prompt: '{prompt}'")
print()
for i, r in enumerate(results, 1):
    print(f"Completion #{i}:")
    print(f"  {r['generated_text']}")
    print()
```

---
## Task 2 — Story Writing

**What it does:** Give the model the start of a story — it continues writing.

Try changing `temperature`:
- Low (0.3) = safe, predictable output
- High (1.2) = creative, surprising output

```python
story_prompt = "Once upon a time, in a small village near the mountains, a young girl discovered"

story = generator(
    story_prompt,
    max_new_tokens=80,
    temperature=2.0,      # 0.0 = very safe/boring, 1.5 = very random/creative
    do_sample=True,       # enables temperature-based sampling
    truncation=True,
)

print(story[0]["generated_text"])
```

---
## Task 3 — Controlled Generation with Parameters

These are the knobs you can turn to control how the model generates text:

| Parameter | What it does | Example |
|---|---|---|
| `max_new_tokens` | How many words to generate | `50` |
| `temperature` | How random/creative the output is | `0.7` |
| `top_k` | Only pick from the top K likely words | `50` |
| `top_p` | Pick from words covering top P% probability | `0.9` |
| `repetition_penalty` | Punish the model for repeating itself | `1.3` |
| `num_return_sequences` | How many different outputs to return | `3` |

```python
prompt = "The best way to learn machine learning is to"

result = generator(
    prompt,
    max_new_tokens=60,
    temperature=0.7,          # balanced creativity
    top_k=50,                 # only consider top 50 words at each step
    top_p=0.9,                # or: pick from words that together cover 90% probability
    repetition_penalty=1.3,   # avoid repeating the same words
    do_sample=True,
    truncation=True,
)

print(f"Prompt: '{prompt}'")
print()
print(result[0]["generated_text"])
```

## Task 4 — See the Model Think: Token Probabilities

**What it does:** Shows the top candidate words at each step — how the decoder decides what to write next.

This is exactly what happens inside GPT-4 at every word position.

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model     = AutoModelForCausalLM.from_pretrained("gpt2")
model.eval()

prompt     = "The capital of France is"
input_ids  = tokenizer(prompt, return_tensors="pt").input_ids

with torch.no_grad():
    outputs = model(input_ids)
    logits  = outputs.logits                      # scores for every word in vocabulary

next_token_logits = logits[0, -1]                 # scores for the NEXT word only
probs = torch.softmax(next_token_logits, dim=-1)  # convert to probabilities
top5  = torch.topk(probs, 5)                      # top 5 candidates

print(f"Prompt: '{prompt}'")
print()
print("Top 5 next-word candidates:")
for token_id, prob in zip(top5.indices.tolist(), top5.values.tolist()):
    word = tokenizer.decode([token_id])
    bar  = "█" * int(prob * 200)
    print(f"  '{word:12s}'  {prob:.2%}  {bar}")
```

---
## Task 5 — Greedy vs Sampling: Two Ways to Pick the Next Word

| Mode | How it picks | Result |
|---|---|---|
| **Greedy** | Always picks the #1 most likely word | Predictable, can get repetitive |
| **Sampling** | Randomly picks from top candidates (using temperature) | More varied and creative |

```python
prompt = "In the future, robots will"

# Greedy — always picks most likely word
greedy = generator(prompt, max_new_tokens=30, do_sample=False, truncation=True)

# Sampling — picks creatively from top candidates
sampled = generator(prompt, max_new_tokens=30, do_sample=True, temperature=1.0, top_k=50, truncation=True)

print("=== GREEDY (always picks most likely word) ===")
print(greedy[0]["generated_text"])
print()
print("=== SAMPLING (creative, picks randomly from top candidates) ===")
print(sampled[0]["generated_text"])
```
